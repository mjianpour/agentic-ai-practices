This is how we simply implement a synchronous program

```python
import time
```

```python
def order_coffee():
    print("coffee ordered")
    time.sleep(3)
    print("enjoy your coffee")
```

```python
start = time.perf_counter()

order_coffee()
order_coffee()
order_coffee()

print(f"total runtime: {time.perf_counter() - start}")
```

And this is how we implement an asynchronous program!

```python
import asyncio
import time
```

```python
async def order_meal():
    print("meal ordered")
    await asyncio.sleep(3)
    print("enjoy your meal")
```

```python
async def do_some_job():
    print("opening the laptop")
    print("waiting to boot!")
    await asyncio.sleep(1.5)
    print("let's work")
```

```python
async def main():
    start = time.perf_counter()

    await asyncio.gather(
        order_meal(),
        do_some_job()
    )

    print(f"total runtime: {time.perf_counter() - start}")
```

```python
await main()
```

Note. Jupyter notebook always already start an asynchoronous loop to empower the notebook. That's why in the last cell we typed "await main()" becaude we're not allowed to run another loop. But if we were in an isolated environment such as a plain .py file, that last cell should have been replaced by asyncio.run(main()) to actually benefit from the privilage of asynchronous programming.
